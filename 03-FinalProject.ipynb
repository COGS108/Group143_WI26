## Research Question

**BROAD RQ:**

Which factors predict the outcome of dogs in the U.S. entering animal shelters, specifically whether a dog is adopted, transferred to another organization, or euthanized?


**SPECIFIC RQs:**
1. Do younger dogs have higher adoption probabilities than older dogs after controlling for breed and intake characteristics?
    > This question uses logistic regression with age at intake as the focal predictor and adoption as the binary outcome (adopted vs. not adopted). Age will be binned into standard shelter-medicine categories (puppy: 0–6 months, juvenile: 6-18 months, young adult: 1.5-3 years, adult: 3-7 years, senior: 7+ years) to account for estimation noise in stray dog age records. Datasets: Austin, Sonoma, and Indiana only. Louisville is excluded due to the absence of an age variable.
2. Do dogs labeled as purebred have higher adoption rates than mixed-breed dogs?
    > Rather than comparing hundreds of inconsistent breed categories, this question collapses breed into a theoretically motivated binary: purebred vs. mixed-breed, which all four datasets can support. Breed labels are visually assigned in shelter environments, so this limitation will be acknowledged. All four datasets (Austin, Sonoma, Indiana, Louisville) will be included in this analysis.
3. Does length of stay itself reduce adoption probability, independent of age and breed?
    > This question uses logistic regression with length of stay as the focal predictor and age and breed as controls. The goal is to isolate a potential shelter effect of asking whether the passage of time in the shelter reduces adoption chances beyond what can be explained by the dog’s characteristics alone. Length of stay will be calculated from intake and outcome dates, which are available across all four datasets.


## Background and Prior Work

The purpose of animal shelters in the United States (U.S.) is to serve as a temporary place for animals to have a safe place to live while they wait for permanent homes. Humans can visit shelters, adopt an animal, and bring it home to care for. However, over the past decade, shelters have been overburdened and overcrowded, leading to many animals being euthanised.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1),<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) More specifically, 11.5% of dogs were euthanized in 2025.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Dogs, in particular, require more human interaction, emphasizing the urgent need to study these species. The longer a dog stays in a shelter, the more likely they are to exhibit stress-related behaviors, such as aggression, which can reduce the likelihood of adoption.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Furthermore, dogs are active animals, and being confined to small spaces for long periods of time can be detrimental to their health<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4).


One way to reduce the number of dogs euthanised is to increase the adoption rate. To do so, we aim to identify the best predictors to increase the likelihood of adoption. In a literature review, researchers identified the most common predictive factors affecting an animal’s adoption rate, including animal type, breed, color, gender, age, location, and length of stay.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Previous literature identified age, size, and color as the strongest predictive factors in a dog’s length of stay.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Other researchers found that age is a strong predictor of adoption, with younger dogs at intake in the shelter more likely to be adopted. Older dogs were more frequently taken in at the animal shelters.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2),<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) Additionally, breed labeling has been shown to influence adoption outcomes, with mixed-breed dogs sometimes facing longer shelter stays than purebred dogs, potentially due to adopter preferences.


Our study focuses on shelters in the U.S., particularly in Austin, Texas; Sonoma, California; Louisville, Kentucky; and Bloomington, Indiana. According to PetFinder data, Texas has the highest rate of dog exports because shelters there cannot support the number of dogs without homes.<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7) California has a medium-to-high level of exported dogs. Kentucky has a medium size of exported dogs, and Indiana has among the lowest rates of exported dogs. We wanted to explore the range of exported dogs based on animal shelter capacity to ensure we captured a diverse set of locations and shelter capacities.
