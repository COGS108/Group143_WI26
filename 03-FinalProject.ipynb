## Abstract

We decided to focus our research on dog adoption because our group consists of dog lovers! One of the most common ways of owning a dog is through adoption, yet shelters are still overcrowded and overburdened. We wanted to further understand if there are any variables that can increase the probability for our furry friends to find homes. With the lack of a national standardized dataset of dog shelter inntakes and outtakes, we looked for shelters that were representative of different regions across the United States, spanning Texas, Kentucky, California, and Indiana. We chose these regions because they span a range of shelter overflow rates, from among the highest in the country (Texas) to among the lowest (Indiana), allowing us to determine whether our findings reflect a local pattern or a broader national trend.

We found that puppies and younger dogs had higher chances of adoption as compared to older, more senior dogs across all shelters we observed. Additionally, mixed dog breeds had higher chances of adoption than purebreds. Lastly, we hypothesized that the longer a dog is in a shelter, the lower their adoption probability. However, we found the opposite: adopted dogs actually had longer median stays than non-adopted dogs, because many non-adopted dogs exit quickly through euthanasia or transfer before ever getting the chance to be adopted. These findings suggest that age and breed are meaningfully associated with adoption outcomes, and that going beyond dog characteristics to focus on shelter practices plays a significant role in which dogs find homes.

## Background and Prior Work

The purpose of animal shelters in the United States (U.S.) is to serve as a temporary place for animals to have a safe place to live while they wait for permanent homes. Humans can visit shelters, adopt an animal, and bring it home to care for. However, over the past decade, shelters have been overburdened and overcrowded, leading to many animals being euthanised.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1),<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) More specifically, 11.5% of dogs were euthanized in 2025.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Dogs, in particular, require more human interaction, emphasizing the urgent need to study these species. The longer a dog stays in a shelter, the more likely they are to exhibit stress-related behaviors, such as aggression, which can reduce the likelihood of adoption.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Furthermore, dogs are active animals, and being confined to small spaces for long periods of time can be detrimental to their health<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4).


One way to reduce the number of dogs euthanised is to increase the adoption rate. To do so, we aim to identify the factors most strongly associated with adoption outcomes. In a meta-analysis, researchers identified the most common factors associated with an animal’s adoption rate, including animal type, breed, color, gender, age, location, and length of stay.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Previous literature identified age, size, and color as the strongest factors associated with a dog’s length of stay.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Other researchers found that age is strongly associated with adoption, with younger dogs at intake in the shelter more likely to be adopted. Older dogs were more frequently taken in at the animal shelters.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2),<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) Additionally, breed labeling has been shown to influence adoption outcomes, with mixed-breed dogs sometimes facing longer shelter stays than purebred dogs, potentially due to adopter preferences.


Our study focuses on shelters in the U.S., particularly in Austin, Texas; Sonoma, California; Louisville, Kentucky; and Bloomington, Indiana. According to PetFinder data, Texas has the highest rate of dog exports because shelters there cannot support the number of dogs without homes.<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7) California has a medium-to-high level of exported dogs. Kentucky has a medium size of exported dogs, and Indiana has among the lowest rates of exported dogs. We wanted to explore the range of exported dogs based on animal shelter capacity to ensure we captured a diverse set of locations and shelter capacities.

### RQ1: Do younger dogs have higher adoption rates?

To test whether age at intake is associated with adoption, we computed each dog's age in months and binned it into five standard shelter-medicine categories: **Puppy** (0–6 months), **Juvenile** (6–18 months), **Young Adult** (1.5–3 years), **Adult** (3–7 years), and **Senior** (7+ years). Louisville was excluded from this analysis because the dataset does not contain an age variable, which leaves Austin, Sonoma, and Bloomington. We then calculated the adoption rate within each age group for every city and used chi-square tests of independence to determine whether the observed differences among age groups were statistically significant regarding adoption rates.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
from scipy import stats
import re

# Using the four cleaned datasets loaded in the Data section above
# (austin, sonoma, bloomington, louisville).
# CSV columns load as strings — convert to datetime so we can subtract dates
# (e.g., Intake Date minus Date Of Birth = age in Sonoma)
for df, date_cols in [
    (austin, ["intake_datetime", "outcome_datetime"]),
    (sonoma, ["Intake Date", "Outcome Date", "Date Of Birth"]),
    (bloomington, ["intakedate", "movementdate"]),
    (louisville, ["indate", "outdate"]),
]:
    for col in date_cols:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors="coerce")

# defining the bins and labels for the age categories
AGE_BINS   = [0, 6, 18, 36, 84, float("inf")]
AGE_LABELS = ["Puppy\n(0–6 mo)", "Juvenile\n(6–18 mo)", "Young Adult\n(1.5–3 yr)",
              "Adult\n(3–7 yr)", "Senior\n(7+ yr)"]

# convert bloomington's open-response age to months
def parse_bloomington_age(age_str):
    if pd.isna(age_str) or str(age_str).strip() in ("nan", ""):
        return np.nan
    total = 0.0
    for m in re.finditer(r"(\d+\.?\d*)\s*(year|month|week|day)", str(age_str).lower()):
        val, unit = float(m.group(1)), m.group(2)
        # map each unit to its equivalent in months
        total += val * {"year": 12, "month": 1, "week": 1/4.345, "day": 1/30.44}[unit]
    return total if total > 0 else np.nan

# filter for dogs only
def build_rq1_subset(df, city, age_col, outcome_col, adopt_label,
                     species_col=None, species_val=None):
    """Filter a dataset to dogs only, extract age in months, classify adoption
    as 1/0, and bin age into the shelter-medicine categories defined above."""
    sub = df.copy()
    if species_col:
        sub = sub[sub[species_col].str.strip().str.upper() == species_val.upper()]
    age = pd.to_numeric(sub[age_col], errors="coerce") # only selecting age column
    adopted = (sub[outcome_col].str.strip().str.upper() == adopt_label.upper()).astype(int)
    out = pd.DataFrame({"city": city, "age_months": age, "adopted": adopted}).dropna(subset=["age_months"])
    out = out[out["age_months"] >= 0]
    # making age bins into discrete age categories
    out["age_category"] = pd.cut(out["age_months"], bins=AGE_BINS,
                                  labels=AGE_LABELS, right=False, include_lowest=True)
    return out

# each dataset stores age differently so here we are coverting each dataset to be consistent for all
#   Austin: age in days → divide by 30.44 to get months
#   Sonoma: has Date Of Birth → subtract from Intake Date to get age in months
#   Bloomington: free-text string → parse with regex function above
rq1 = pd.concat([
    build_rq1_subset(austin.assign(age_mo=austin["age_upon_intake_(days)"] / 30.44),
                     "Austin", "age_mo", "outcome_type", "Adoption", "animal_type", "Dog"),
    build_rq1_subset(sonoma.assign(age_mo=(sonoma["Intake Date"] - sonoma["Date Of Birth"]).dt.total_seconds() / (3600*24*30.44)),
                     "Sonoma", "age_mo", "Outcome Type", "ADOPTION", "Type", "DOG"),
    build_rq1_subset(bloomington.assign(age_mo=bloomington["animalage"].apply(parse_bloomington_age)),
                     "Bloomington", "age_mo", "movementtype", "Adoption", "speciesname", "Dog"),
], ignore_index=True)

# --- Grouped bar chart: adoption rate by age category, one bar per city ---
CITY_ORDER = ["Austin", "Sonoma", "Bloomington"]
AGE_LABELS_FLAT = [l.replace("\n", " ") for l in AGE_LABELS]
CITY_COLORS = {"Austin": "#ff9a87", "Sonoma": "#5B9BD5", "Bloomington": "#77C28D"}

rates = rq1.groupby(["city", "age_category"], observed=True)["adopted"].agg(["mean", "count"]).reset_index()
rates["se"] = np.sqrt(rates["mean"] * (1 - rates["mean"]) / rates["count"])

fig, ax = plt.subplots(figsize=(13, 6))
x = np.arange(len(AGE_LABELS))
n_cities = len(CITY_ORDER)
w = 0.25

for i, city in enumerate(CITY_ORDER):
    sub = rates[rates["city"] == city].set_index("age_category").reindex(AGE_LABELS)
    bars = ax.bar(x + i * w, sub["mean"], w, yerr=sub["se"] * 1.96,
                  label=city, color=CITY_COLORS[city], edgecolor="white", capsize=3, alpha=0.85)
    for j, (rate, n) in enumerate(zip(sub["mean"], sub["count"])):
        ax.text(x[j] + i * w, rate + sub["se"].iloc[j] * 1.96 + 0.01,
                f"{rate:.0%}", ha="center", va="bottom", fontsize=9, fontweight="bold")

ax.set_xticks(x + w)
ax.set_xticklabels(AGE_LABELS_FLAT, fontsize=12)
ax.set_xlabel("Age Category at Intake", fontsize=14)
ax.set_ylabel("Adoption Rate", fontsize=14)
ax.set_ylim(0, 0.85)
ax.yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1, decimals=0))
ax.tick_params(axis="y", labelsize=12)
ax.set_title("Dogs' Adoption Rate by Age Category (Per City)",
             fontsize=16, fontweight="bold")
ax.legend(title="City", fontsize=11)
ax.yaxis.grid(True, linestyle="--", alpha=0.5)
ax.set_axisbelow(True)
sns.despine()
plt.tight_layout()
plt.savefig("results/rq1_adoption_by_age.png", dpi=150, bbox_inches="tight")
plt.show()

# --- Chi-square test on the pooled data ---
# H0: age category and adoption outcome are independent (no association)
# A small p-value (< 0.05) means we reject H0 and conclude they are associated
ct_pooled = pd.crosstab(rq1["age_category"], rq1["adopted"])
chi2, p, dof, _ = stats.chi2_contingency(ct_pooled)
print(f"Pooled Chi-Square Test:  χ² = {chi2:.2f},  df = {dof},  p = {p:.2e}")
print(f"  → {'Significant' if p < 0.05 else 'Not significant'} (p {'<' if p < 0.05 else '>='} 0.05)")

# Confirm the pattern holds within each city individually
print("\nIndependent chi-Square tests for each city (consistency check):")
print("-" * 50)
for city in CITY_ORDER:
    ct = pd.crosstab(rq1[rq1["city"]==city]["age_category"], rq1[rq1["city"]==city]["adopted"])
    chi2, p, dof, _ = stats.chi2_contingency(ct)
    print(f"  {city}:  χ² = {chi2:.2f},  df = {dof},  p = {p:.2e}  → {'Significant' if p < 0.05 else 'Not significant'}")
```

**RQ1 Findings:**

When combining data across Austin, Sonoma, and Bloomington, adoption rates decline steadily with age. Puppies have the highest adoption rate, followed by juveniles and young adults, while seniors have the lowest. A chi-square test on the combined data confirms that this association between age category and adoption outcome is statistically significant (*p* < .001), supporting our first hypothesis that younger dogs are more likely to be adopted.

This pattern is representative of only a single city, but rather three cities. Chi-square tests of each city show the relationship between age and adoption rates is independently significant in Austin, Sonoma, and Bloomington (all *p* < .001), which strengthens our confidence that the finding generalizes across shelters with different capacities and regions.

One limitation is that the data quality of age varies across datasets. Austin records age in days (which is highly precise), Sonoma computes it from a date-of-birth field that may be estimated for strays, and Bloomington uses open-response entries (e.g.,"2 years"), which introduces some guessing.

### RQ2: Do purebred dogs have higher adoption rates than mixed-breed dogs?

To test whether giving dogs labels for possible breed classification influences adoption, we classified each dog as **Purebred** or **Mixed** based on whether its breed field contained indicators such as "Mix," "/", or "Cross." In other words, we've collapsed the breed types into binary bins so that we avoid comparing hundreds of inconsistent breed labels. Furthermore, this binary approach allows us to investigate the stigma between purebred preference and mixed-breed. All four cities are included in this analysis. We computed adoption rates by breed type per city and tested for statistical association using chi-square tests.

```python
# classify breed as Mixed or Purebred using keyword heuristics.
# in other words, if the breed string contains "Mix", "/", or "Cross" it is labeled Mixed.
def classify_breed(breed_str):
    """Classify a breed string as 'Mixed' or 'Purebred'."""
    if pd.isna(breed_str):
        return "Unknown"
    b = breed_str.upper()
    return "Mixed" if ("MIX" in b or "/" in b or "CROSS" in b) else "Purebred"

def build_rq2_subset(df, city, breed_col, outcome_col, adopt_label,
                     species_col=None, species_val=None):
    """Filter to dogs, apply the breed classifier, and create a binary adopted column."""
    sub = df.copy()
    if species_col:
        sub = sub[sub[species_col] == species_val]
    return pd.DataFrame({
        "city": city,
        "breed_type": sub[breed_col].apply(classify_breed),
        "adopted": (sub[outcome_col] == adopt_label).astype(int)
    })

# combine into one dataframe
rq2 = pd.concat([
    build_rq2_subset(austin, "Austin", "breed", "outcome_type", "Adoption", "animal_type", "Dog"),
    build_rq2_subset(louisville, "Louisville", "breed", "outtype", "ADOPTION", "animaltype", "DOG"),
    build_rq2_subset(sonoma, "Sonoma", "Breed", "Outcome Type", "ADOPTION", "Type", "DOG"),
    build_rq2_subset(bloomington, "Bloomington", "breedname", "movementtype", "Adoption", "speciesname", "Dog"),
], ignore_index=True)

# drop rows where breed could not be determined
rq2 = rq2[rq2["breed_type"] != "Unknown"]

# visualization for RQ2
CITY_ORDER_4 = ["Austin", "Louisville", "Sonoma", "Bloomington"]
BREED_COLORS = {"Mixed": "#5B9BD5", "Purebred": "#ED7D31"}

fig, ax = plt.subplots(figsize=(10, 5))

rates = rq2.groupby(["city", "breed_type"])["adopted"].agg(["mean", "count"]).reset_index()
rates["se"] = np.sqrt(rates["mean"] * (1 - rates["mean"]) / rates["count"])

x = np.arange(len(CITY_ORDER_4))
w = 0.35  
for i, bt in enumerate(["Mixed", "Purebred"]):
    sub = rates[rates["breed_type"] == bt].set_index("city").reindex(CITY_ORDER_4)
    bars = ax.bar(x + i*w, sub["mean"], w, yerr=sub["se"]*1.96,
                  label=bt, color=BREED_COLORS[bt], edgecolor="white", capsize=4, alpha=0.85)
    for j, (rate, n) in enumerate(zip(sub["mean"], sub["count"])):
        ax.text(x[j] + i*w, rate + sub["se"].iloc[j]*1.96 + 0.02,
                f"{rate:.1%}\n(n={int(n):,})", ha="center", va="bottom", fontsize=9.5)

ax.set_xticks(x + w/2)
ax.set_xticklabels(CITY_ORDER_4, fontsize=12)
ax.set_ylabel("Adoption Rate", fontsize=14)
ax.set_ylim(0, 0.85)
ax.yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1, decimals=0))
ax.tick_params(axis="y", labelsize=12)
ax.set_title("Dogs' Adoption Rate by Breed Type (All Four Cities)",
             fontsize=16, fontweight="bold")
ax.legend(title="Breed Type", fontsize=11)
ax.yaxis.grid(True, linestyle="--", alpha=0.5)
ax.set_axisbelow(True)
sns.despine()
plt.tight_layout()
plt.savefig("results/rq2_adoption_by_breed.png", dpi=150, bbox_inches="tight")
plt.show()

# chi-square tests of independence
print("\nChi-Square Tests: Breed Type vs. Adoption Outcome")
print("=" * 60)
for city in CITY_ORDER_4:
    ct = pd.crosstab(rq2[rq2["city"]==city]["breed_type"], rq2[rq2["city"]==city]["adopted"])
    chi2, p, dof, _ = stats.chi2_contingency(ct)
    print(f"\n{city}:  χ² = {chi2:.2f},  df = {dof},  p = {p:.2e}")
    print(f"  → {'Significant' if p < 0.05 else 'Not significant'} (p {'<' if p < 0.05 else '>='} 0.05)")
```

**RQ2 Findings:**

The results did not support our hypothesis, as mixed-breed dogs are adopted at *higher* rates than purebred dogs in every city. In Austin, 46% of mixed-breed dogs are adopted versus 33% of purebred dogs. Louisville shows a similar pattern (40% vs. 33%). Sonoma and Bloomington follow the same direction, with mixed-breed adoption rates of 28% and 59% compared to purebred rates of 18% and 53%, respectively.

Chi-square tests confirm that the association between breed type and adoption is statistically significant in all four cities (all *p* < 0.001). This means the difference is unlikely due to chance. One major reason why the results differed from our hypothesis is that mixed-breed dogs greatly outnumber purebred dogs in the shelters we examined. It is also possible that adopters visiting shelters are specifically seeking mixed-breed dogs rather than going to breeders for purebreds, though it is more likely the former reason.

One limitation is that breed labels are assigned by shelter staff based on their subjective assessments, not genetic testing. Some dogs classified as purebred may actually be mixed, and vice versa. This classification noise is applied consistently across all datasets, but it may attenuate or exaggerate the true effect.

### RQ3: Is length of stay associated with adoption outcome?

To test whether time spent in a shelter is associated with adoption outcome, we computed each dog's length of stay (LOS) in days from intake and outcome dates. All four cities are included. We compared the median LOS for adopted vs. not-adopted dogs using box plots, investigated how adoption rates change across LOS ten bins using a line plot, and tested for distributional differences using the Mann-Whitney U test (a non-parametric test chosen because LOS is heavily right-skewed in all cities).

```python
def build_rq3_subset(df, city, los_col, outcome_col, adopt_label,
                     species_col=None, species_val=None):
    """Filter to dogs, extract length of stay in days, classify adoption as 1/0,
    and remove invalid LOS values (negative or > 365 days)."""
    sub = df.copy()
    if species_col:
        sub = sub[sub[species_col].str.strip().str.upper() == species_val.upper()]
    out = pd.DataFrame({
        "city": city,
        "length_of_stay_days": pd.to_numeric(sub[los_col], errors="coerce"),
        "adopted": (sub[outcome_col].str.strip().str.upper() == adopt_label.upper()).astype(int)
    })
    # keep only rows with valid LOS between 0 and 365 days
    return out[(out["length_of_stay_days"].notna()) &
               (out["length_of_stay_days"] >= 0) &
               (out["length_of_stay_days"] <= 365)]

# combine LOS into one dataframe
rq3 = pd.concat([
    build_rq3_subset(austin, "Austin", "length_of_stay_days", "outcome_type", "Adoption", "animal_type", "Dog"),
    build_rq3_subset(louisville, "Louisville", "length_of_stay_days", "outtype", "ADOPTION", "animaltype", "DOG"),
    build_rq3_subset(sonoma, "Sonoma", "Days in Shelter", "Outcome Type", "ADOPTION", "Type", "DOG"),
    build_rq3_subset(bloomington, "Bloomington", "length_of_stay_days", "movementtype", "Adoption", "speciesname", "Dog"),
], ignore_index=True)

# create a human-readable label for plotting
rq3["outcome_label"] = rq3["adopted"].map({1: "Adopted", 0: "Not Adopted"})

# visualization for RQ3
CAP_DAYS = 120
PALETTE  = {"Adopted": "#5B9BD5", "Not Adopted": "#ED7D31"}
fig, axes = plt.subplots(2, 2, figsize=(12, 9), sharey=False)
fig.suptitle("Dogs' Length of Stay by Adoption Outcome\n(capped at 120 days for readability)",
             fontsize=17, fontweight="bold", y=1.01)

for ax, city in zip(axes.flat, CITY_ORDER_4):
    sub = rq3[(rq3["city"] == city) & (rq3["length_of_stay_days"] <= CAP_DAYS)]

    med_a = sub[sub["adopted"]==1]["length_of_stay_days"].median()
    med_n = sub[sub["adopted"]==0]["length_of_stay_days"].median()
    n_a   = (sub["adopted"]==1).sum()
    n_n   = (sub["adopted"]==0).sum()

    sns.boxplot(data=sub, x="outcome_label", y="length_of_stay_days", hue="outcome_label",
                order=["Adopted","Not Adopted"], palette=PALETTE, width=0.45, legend=False, ax=ax,
                flierprops=dict(marker="o", markersize=2, alpha=0.25))

    for i, (med, label) in enumerate([(med_a,"Adopted"), (med_n,"Not Adopted")]):
        ax.text(i, med + 1.5, f"med={med:.0f}d", ha="center", va="bottom", fontsize=10.5, fontweight="bold")

    ax.set_title(city, fontsize=14, fontweight="bold")
    ax.set_xlabel("")
    ax.set_ylabel("Days in Shelter", fontsize=13)
    ax.set_xticklabels([f"Adopted\n(n={n_a:,})", f"Not Adopted\n(n={n_n:,})"], fontsize=11)
    ax.tick_params(axis="y", labelsize=11)
    sns.despine(ax=ax)

plt.tight_layout()
plt.savefig("results/rq3_los_boxplots.png", dpi=150, bbox_inches="tight")
plt.show()

print("\nMedian Length of Stay (days) by City and Outcome:")
print(rq3.groupby(["city","outcome_label"])["length_of_stay_days"].median().unstack().round(1))

# Mann-Whitney U tests (non-parametric)
# We use Mann-Whitney instead of a t-test because LOS is heavily right-skewed
# (most dogs leave quickly, but some stay for months)
print("\nMann-Whitney U Tests: LOS — Adopted vs. Not Adopted (two-sided)")
print("=" * 65)
for city in CITY_ORDER_4:
    a_los = rq3[(rq3["city"]==city) & (rq3["adopted"]==1)]["length_of_stay_days"]
    n_los = rq3[(rq3["city"]==city) & (rq3["adopted"]==0)]["length_of_stay_days"]
    u_stat, p = stats.mannwhitneyu(a_los, n_los, alternative="two-sided")
    print(f"\n{city}:  Adopted median = {a_los.median():.1f}d (n={len(a_los):,}),  "
          f"Not Adopted median = {n_los.median():.1f}d (n={len(n_los):,})")
    print(f"  U = {u_stat:.0f},  p = {p:.2e}")
    print(f"  → {'Significant' if p < 0.05 else 'Not significant'} (p {'<' if p < 0.05 else '>='} 0.05)")
```

**RQ3 Findings:**

Across all four cities, dogs that were adopted had a *longer* median length of stay than dogs that were not adopted. In Austin, adopted dogs had a median stay of 7 days versus 3 days for non-adopted dogs. Louisville showed a similar pattern (10 vs. 5 days). Sonoma had the largest gap (28 vs. 3 days), and Bloomington fell in between (19 vs. 8 days). Mann-Whitney U tests confirmed that these distributional differences are statistically significant in all four cities (all *p* < 0.001).

This result runs in opposition to our original hypothesis that longer stays reduce adoption probability. Instead, the pattern reflects a confound: many non-adopted dogs exit the shelter very quickly through euthanasia or transfer, which actually shortens their length of stay. Adopted dogs, on the other hand, need time in the shelter to be seen and selected by adopters. In other words, a short stay does not mean a dog was easy to adopt. Rather, it may mean the dog was removed from the shelter through a non-adoption pathway before it had the chance.

This finding highlights a major limitation of interpreting length of stay as a straightforward indicator of adoptability. Dogs may be held longer *because* shelters are actively trying to place them, while sadly, dogs deemed less adoptable may be euthanized or transferred sooner.

## Discussion and Conclusion

In this project, we explored how age, breed, and length of stay affect dog adoption outcomes across four animal shelters in the U.S. Overall, age had the strongest and most consistent impact on adoption. Across multiple cities, we saw a clear pattern where adoption rates steadily decreased as dogs got older. Puppies had the highest adoption rates, while senior dogs had the lowest. This supports our original hypothesis and suggests that adopters strongly prefer younger dogs, likely because they are seen as easier to train and have longer lifespans.

Our findings for breed were unexpected. We originally thought that purebred dogs would be adopted more often, but the data showed that mixed-breed dogs had higher adoption rates in every city. This could be because mixed breeds make up a larger portion of shelter populations, giving them more opportunities to be adopted. It may also reflect changing attitudes, where adopters are more open to mixed breeds or believe they are healthier. Another possibility is that breed labeling in shelters is not always accurate, which could affect how this variable is represented in the data.

For length of stay, we were not able to find a clear relationship with adoption probability. While we expected that longer stays would decrease the likelihood of adoption, our analysis showed that other factors, such as transfers or euthanasia, can shorten a dog’s stay in ways that are not related to adoption. Because of this, length of stay alone was not a reliable indicator of adoptability in our study. This highlights how shelter processes can affect the data and make certain variables harder to interpret.

Overall, these results show that adoption outcomes are influenced by multiple factors, but not all of them behave as expected. Age is a strong and consistent factor, while breed and length of stay are more complex and influenced by both data limitations and real-world shelter practices. By comparing multiple shelters across different regions, we were able to see patterns that were consistent as well as identify areas where results were less clear.

There are several limitations to this project. The datasets from different shelters were not fully standardized, and some data may be missing or inconsistent. Breed classification is based on visual estimates rather than confirmed information, which can introduce error. We also simplified breed into just two categories, which does not capture the full diversity of dogs. In addition, we grouped all non-adoption outcomes together, even though transfers and euthanasia are very different. We also did not include other important factors like health, behavior, or shelter resources, which likely play a significant role in adoption outcomes. Because of these limitations, our results show patterns but do not prove cause and effect.

In the future, this project could be improved by using more detailed and standardized datasets, separating different types of outcomes, and including additional variables, such as a national standard for accounting for various shelters and the types of data they collect. More advanced models could also help better understand how multiple factors interact at the same time. Overall, our findings show that younger and mixed-breed dogs have higher chances of adoption, while older and purebred dogs may face more challenges. Understanding these patterns can help shelters make better decisions and raise awareness so that all dogs have a better chance at finding their "fur"-ever home.
